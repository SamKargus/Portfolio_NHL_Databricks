```sql
-- Game - Find Game ID
-- Lists games so you can copy a game_id into the other Game queries.
-- Change the values in params only; everything below reads from it.
WITH params AS (                              -- change settings here only
  SELECT
    'MTL' AS team,                      -- team abbreviation; CAST(NULL AS STRING) = every team
    20252026 AS season,                 -- e.g. 20242025; CAST(NULL AS INT) = every season
    2 AS game_type,                     -- 2 = regular season, 3 = playoffs
    CAST(NULL AS STRING) AS opponent,   -- opponent abbreviation, e.g. 'TOR'; NULL = every opponent
    CAST(NULL AS DATE) AS game_date     -- e.g. DATE'2025-10-08'; NULL = every date
)
SELECT g.game_id,
       g.game_date,
       a.abbrev                 AS away,
       g.away_final_score       AS away_score,
       h.abbrev                 AS home,
       g.home_final_score       AS home_score,
       g.last_period_type       AS decided_in,
       g.game_state,
       g.venue
FROM nhl.gold.dim_game g
JOIN nhl.gold.dim_team h ON h.team_id = g.home_team_id
JOIN nhl.gold.dim_team a ON a.team_id = g.away_team_id
CROSS JOIN params pr
WHERE g.game_type = pr.game_type
  AND (pr.season IS NULL OR g.season = pr.season)
  AND (pr.team IS NULL OR pr.team IN (h.abbrev, a.abbrev))
  AND (pr.opponent IS NULL OR pr.opponent IN (h.abbrev, a.abbrev))
  AND (pr.game_date IS NULL OR g.game_date = pr.game_date)
ORDER BY g.game_date, g.game_id;
```
