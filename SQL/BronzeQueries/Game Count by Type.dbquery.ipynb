```sql
SELECT
    game_type,
    CASE game_type
      WHEN 1  THEN 'Preseason'
      WHEN 2  THEN 'Regular season'
      WHEN 3  THEN 'Playoffs'
      WHEN 4  THEN 'All-Star'
      WHEN 6  THEN 'World Cup group stage'
      WHEN 7  THEN 'World Cup knockout stage'
      WHEN 8  THEN 'World Cup pre-tournament'
      WHEN 9  THEN 'Olympics'
      WHEN 10 THEN 'Young Stars'
      WHEN 12 THEN 'All-Stars exhibition'
      WHEN 13 THEN '2004 labor dispute lost games'
      WHEN 14 THEN 'Canada Cup'
      WHEN 18 THEN 'Overseas exhibition'
      WHEN 19 THEN 'Four Nations Face-Off'
      ELSE 'Unknown'
    END                     AS game_type_name,
    count(*)                AS games,
    min(game_date)          AS first_game,
    max(game_date)          AS last_game,
    any_value(matchup)      AS sample_matchup
  FROM (
    SELECT
      CAST(get_json_object(raw_json, '$.gameType') AS INT)  AS game_type,
      get_json_object(raw_json, '$.gameDate')               AS game_date,
      concat(get_json_object(raw_json, '$.awayTeam.abbrev'), ' @ ',
             get_json_object(raw_json, '$.homeTeam.abbrev')) AS matchup
    FROM nhl.bronze.nhl_events
  )
  GROUP BY game_type
  ORDER BY game_type
```
